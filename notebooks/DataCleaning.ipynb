**Limpieza de Datos**

Para emepezar tenemos que cargar y limpiar la base de datos usando al libreria de Pandas.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/ensanut2022_muestras.csv", sep=";")
df.head()
```

Empezamos por imprimir todas las columnas para luego escoger con cuales nos queremos quedar

```python
for col in df.columns:
    print(col)
```

```python
wanted_columns = ["FOLIO_I", "entidad", "desc_ent", "municipio", 
                  "desc_mun", "h0302", "h0303", "valor_COL_HDL", 
                  "valor_COL_LDL", "valor_COLEST", "valor_GLU_SUERO", 
                  "valor_TRIG", "valor_EAG"]

numeric_cols = ["valor_COL_HDL", "valor_COL_LDL", "valor_COLEST", 
                  "valor_GLU_SUERO", "valor_TRIG", "valor_EAG"]
data = df[wanted_columns]
```

Ahora quitamos todas las filas que no tienen valores y cambiamos de tipo de datos

```python
data[numeric_cols] = data[numeric_cols].replace({' ': pd.NA, ' ': pd.NA, ',': '.'})
```

```python
data = data.dropna()
```

```python
data.info()
```

```python
for col in numeric_cols:
    data[col] = data[col].str.replace(',', '.')
```

```python
data[numeric_cols] = data[numeric_cols].astype('float')
```

```python
data.info()
```

```python
data.describe()
```

Vamos a renombrar algunas de las columnas para poder entenderlo de manera mas facil

```python
data = data.rename(columns={'h0302': 'sexo', 'h0303': 'edad'})
```

```python
data
```

```python
data = data.reset_index().drop(columns=['index'])
```

Finalmente guardamos los datos limpios para poder trabajar con ellos de manera mas facil

```python
data.to_csv('../data/clean_data.csv')
```
